```python
# creating a new column  Age-group in the datase
df["Age_group"] = df["Age"].apply(
    lambda x: "0-18" if x <= 18 else
    "19-30" if x <= 30 else
    "31-45" if x <= 45 else
    "46-60" if x <= 60 else
    "61-75" if x <= 75 else
    "76-85"
)
```

```python
# which age group has highest number of admission
df["Age_group"].value_counts()
```

```python
#How do hospital admissions change monthly and yearly?
#monthely change
df['Date of Admission']=pd.to_datetime(df['Date of Admission'])
monthely_change=df.groupby(df["Date of Admission"].dt.month).size()
print('Hospital Admission change Monthely',monthely_change)
```

```python
monthely_change.plot(
    kind="line",
    marker="o",
    figsize=(8, 5),
    color="teal"
)

plt.title("Monthly Hospital Admissions Trend")
plt.xlabel("Month")
plt.ylabel("Total Admissions")
plt.xticks(rotation=45)
plt.grid(True)
plt.show()
```

```python
#yearley wise 
yearly_wise=df.groupby(df['Date of Admission'].dt.year).size()
print('Hospital Admission change Year wise',yearly_wise)
```

```python
yearly_wise.plot(
    kind="line",
    marker="o",
    figsize=(8, 5),
    color="teal"
)

plt.title("Yearly Hospital Admissions Trend")
plt.xlabel("Year")
plt.ylabel("Total Admissions")
plt.xticks(rotation=45)
plt.grid(True)
plt.show()
```

```python
# changing data type of the column with date and time format
df['Discharge Date']=pd.to_datetime(df['Discharge Date'], format="%d-%m-%Y")
```

```python
# creating a nuw column patients stay 
df['Patient_stay']=df['Date of Admission']-df['Discharge Date']
```

```python
#Which hospitals have the longest average patient stay?
hospital_stay = df.groupby("Hospital")["Patient_stay"].mean()

hospital_stay = hospital_stay.sort_values(ascending=False)

print(hospital_stay.head(10))
```

```python
avg_billing = df.groupby("Admission Type")["Billing Amount"].mean()

print(avg_billing.round(2))
```

```python
#How does average billing differ by admission type?
print("Highest average billing:", avg_billing.idxmax())
print("Average billing amount:", round(avg_billing.max(), 2))
```
